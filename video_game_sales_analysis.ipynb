## 2. Confidence Interval Analysis: Proportion of 3rd Person Games

### Research Question
What is the proportion of 3rd person video games in the population?

### Dataset Cleaning

```python
# Check for missing values in ThirdPerson column
missing_count = (df["ThirdPerson"].isna()).sum()
print(f"Missing values in ThirdPerson: {missing_count}")
print(f"Unique values: {df['ThirdPerson'].unique()}")
```

### Descriptive Statistics

```python
# Calculate proportion of 3rd person games
sample_size = len(df)
print(f"Sample size: {sample_size}")

proportions = df["ThirdPerson"].value_counts(normalize=True)
print(f"\nProportions by perspective:")
print(f"3rd Person (1): {proportions[1]:.2%}")
print(f"Other Perspective (0): {proportions[0]:.2%}")
```

### Confidence Interval Calculation

We construct a 90% confidence interval using bootstrap resampling. This means we examine quantiles 0.05 and 0.95 of the sampling distribution.

```python
# Generate bootstrap sampling distribution
np.random.seed(42)
sampled_proportions = []

for i in range(1000):
    sample = df["ThirdPerson"].sample(len(df), replace=True)
    proportion_third_person = sample.mean()
    sampled_proportions.append(proportion_third_person)

df_sample = pd.DataFrame({'proportion': sampled_proportions})

# Calculate confidence interval bounds
lower_bound = df_sample["proportion"].quantile(0.05)
upper_bound = df_sample["proportion"].quantile(0.95)

print(f"90% Confidence Interval for Proportion of 3rd Person Games:")
print(f"Lower Bound: {lower_bound:.4f} ({lower_bound:.2%})")
print(f"Upper Bound: {upper_bound:.4f} ({upper_bound:.2%})")
```

```python
# Visualize the confidence interval
plt.figure(figsize=(10, 6))
plt.hist(df_sample['proportion'], bins=30, alpha=0.7, edgecolor='black', color='steelblue')
plt.axvline(lower_bound, color='red', linestyle='--', linewidth=2.5, label=f'Lower Bound: {lower_bound:.4f}')
plt.axvline(upper_bound, color='red', linestyle='--', linewidth=2.5, label=f'Upper Bound: {upper_bound:.4f}')
plt.axvline(0.75, color='green', linestyle='-', linewidth=2.5, label='Sample Proportion: 0.75')
plt.xlabel('Proportion of 3rd Person Games', fontsize=11)
plt.ylabel('Frequency', fontsize=11)
plt.title('Sampling Distribution of 3rd Person Game Proportion (90% CI)', fontsize=12, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Confidence Interval Interpretation

**90% Confidence Interval: [73%, 77%]**

We are 90% confident that the true proportion of all video games (in the population) that are played in third person lies between 73% and 77%. This means that if we were to repeatedly sample from the population and construct a 90% confidence interval from each sample, approximately 90% of those intervals would contain the true population proportion.

The confidence level of 90% indicates our willingness to accept a 10% chance that our interval does not contain the true parameter. This is a standard trade-off between precision (narrower intervals) and confidence (higher likelihood of capturing the true value).

## 3. Hypothesis Testing Analysis: Sales Impact of 3rd Person Perspective

### Research Question
Do video games with a third-person perspective have significantly higher average sales than games with other perspectives?

### Dataset Cleaning
Selected variables: ThirdPerson and US Sales (millions)

```python
# Check for missing values
print("Missing values check:")
print(f"US Sales (millions): {(df['US Sales (millions)'].isna()).sum()}")
print(f"ThirdPerson: {(df['ThirdPerson'].isna()).sum()}")

# Verify data types
print(f"\nUnique values in ThirdPerson: {df['ThirdPerson'].unique()}")
```

### Data Limitations
- **Geographic Scope**: This dataset only includes US sales, not global sales, so results may not reflect international market performance.
- **Confounding Variables**: Other factors affecting sales (marketing budgets, genre, release timing, platform popularity) are not controlled for in this analysis.
- **Temporal Scope**: Data is limited to 2004-2010 and may not reflect current gaming trends.
- **Skewed Distribution**: Sales data exhibits significant right skew with outliers that may inflate the mean.

### Descriptive Analysis

```python
# Summary statistics for sales
print("US Sales (millions) Summary Statistics:")
print(df["US Sales (millions)"].describe())

# Separate sales by perspective
third_person_sales = df[df['ThirdPerson'] == 1]['US Sales (millions)']
non_third_person_sales = df[df['ThirdPerson'] == 0]['US Sales (millions)']

print(f"\n3rd Person Games (n={len(third_person_sales)}):")
print(f"  Mean: ${third_person_sales.mean():.2f}M, Median: ${third_person_sales.median():.2f}M")

print(f"\nNon-3rd Person Games (n={len(non_third_person_sales)}):")
print(f"  Mean: ${non_third_person_sales.mean():.2f}M, Median: ${non_third_person_sales.median():.2f}M")
```

### Hypothesis Test Setup

**Null Hypothesis (H₀):** The difference between median sales of 3rd person games and non-3rd person games is zero.
- H₀: M₃ₚₚ - M_other = 0

**Alternative Hypothesis (H₁):** 3rd person games have significantly higher median sales.
- H₁: M₃ₚₚ - M_other > 0

**Significance Level:** α = 0.05
This means we require strong evidence (p-value < 0.05) to reject the null hypothesis and accept only a 5% chance of Type I error.

**Test Statistic:** Difference in sample medians obtained through bootstrap resampling

**Rationale for Using Median:** Due to the right-skewed distribution of sales, the median is a more robust measure of central tendency than the mean, as it is less influenced by extreme outliers.

```python
# Conduct hypothesis test via bootstrap resampling
np.random.seed(42)
simulated_diffs = []

for i in range(1000):
    resampled_third = third_person_sales.sample(len(third_person_sales), replace=True)
    resampled_non_third = non_third_person_sales.sample(len(non_third_person_sales), replace=True)
    simulated_diff = resampled_third.median() - resampled_non_third.median()
    simulated_diffs.append(simulated_diff)

sampling_dist = pd.DataFrame({'difference': simulated_diffs})
observed_diff = third_person_sales.median() - non_third_person_sales.median()

# Calculate p-value
p_value = (sampling_dist['difference'] >= observed_diff).sum() / len(sampling_dist)

print(f"Hypothesis Test Results:")
print(f"Observed Difference in Medians: ${observed_diff:.4f}M")
print(f"P-value: {p_value:.4f}")
print(f"\nConclusion: {'Reject H₀' if p_value < 0.05 else 'Fail to Reject H₀'} (α = 0.05)")
```

### Hypothesis Test Interpretation

**Result:** With a p-value of 1.0, which is far greater than our significance level of α = 0.05, we **fail to reject the null hypothesis**.

**Interpretation:** There is insufficient statistical evidence to support the claim that 3rd person video games have significantly higher median sales than non-3rd person video games. The observed difference of $0.06M in medians could easily have occurred by random chance.

**Statistical Significance vs. Practical Significance:** Even if the difference were statistically significant, the magnitude ($0.06M) is negligible compared to the overall sales distribution, suggesting that game perspective alone does not drive commercial success.

**Key Takeaway:** The perspective of a video game does not significantly predict its sales performance. Other factors such as genre, marketing budget, platform, and brand reputation likely play more important roles in determining video game sales.
